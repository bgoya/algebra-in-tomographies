a) Se considera un *problema inverso* porque el problema no se define en cómo medir mejor al objeto, sino en cómo hacer las mediciones obtenidas más acertadas al escaneo ideal desconocido.

Se parte de una medición conocida $\hat{b}$, que se obtiene tras escanear un vector $\hat{x}$ desconocido con la matriz de direcciones $A$ preestablecida, dando lugar al sistema $A\hat{x} = \hat{b}$. Vimos que tanto $\hat{x}$ como $\hat{b}$ contienen errores de medición imperceptibles por el escáner, los cuales vamos a querer suprimir al máximo para recuperar de manera más fiable el verdadero escaneo $\vec{b}$ del sistema $A\vec{x} = \vec{b}$.

b) Como vimos, el $rango(A)$ es el que nos va a decir si las reconstrucciones son uno a uno porque, bajo su Teorema del Rango, si $rango(A) \neq n$, entonces no va a haber una solución única para los sistemas $A\vec{x} = \vec{b}$. Como vimos antes, van a haber $\vec{v} \in Nul(A)$ y $\vec{v} \neq \vec{0}$ tales que cumplen $A(\vec{x} + \vec{v}) = A\vec{x} + A\vec{v} = \vec{b} + 0 = \vec{b}$, es decir, el escáner da con dos mediciones visualmente diferentes pero matemáticamente idénticas.

c) El ruido afecta a $\vec{b}$, ya que este pasa a ser una expresión tipo $\vec{b} = \vec{p} + \vec{r}$, donde $\vec{p} \in Col(A_7)$ es la proyección ortogonal de $\vec{b}$ sobre $Col(A_7)$ y $\vec{r}$ es el residuo ortogonal. La resolución por mínimos cuadrados nos devuelve la recta que mejor aproxima a los datos dados. El residuo nos da la diferencia entre la aproximación, esto es, la evaluación en la recta de mínimos cuadrados, contra el valor verdadero. Va a ser suficiente si y solo si el $rango(A) = n$ porque, de no ser así, ya vimos que van a haber más de una posible solución al sistema y, por ende, aproximaciones que no son las que mejor representan a la imágen pero son matemáticamente correctas.

 La regularización supone que la imagen es relativamente suave, es decir, que los píxeles vecinos tienen valores parecidos, salvo en algunos bordes. El término |Dx|2 penaliza las diferencias entre píxeles vecinos y, como D1=0, una imagen constante no recibe penalización. Esto ayuda a reducir el ruido, ya que el ruido genera variaciones bruscas entre píxeles.

Como se observa en la figura final del ejercicio 10, para =0 se mantiene el granulado (|Dx|25, residuo 46, frente a un ruido de 76). Para =35, se reduce el granulado y se conservan los detalles finos (|Dx|14.5, residuo 76). En cambio, para =100, la imagen queda demasiado suavizada (|Dx|11.4, residuo 104). Esto ocurre porque la penalización cuadrática favorece repartir los saltos grandes entre píxeles, difuminando los bordes reales.

### Integración de conceptos

**Problema inverso → modelo de reconstrucción.** El problema se plantea como $b=Ax$, donde $x$ representa la imagen, $A$ describe las proyecciones y $b$ contiene las mediciones. Es un problema inverso porque conocemos las mediciones y buscamos recuperar la imagen.

**Espacio nulo → unicidad de la solución.** La reconstrucción es única si $\operatorname{Nul}(A)=\{0\}$, equivalente a que $\operatorname{rango}(A)=n^2$. Si existe un vector $v\neq 0$ tal que $Av=0$, entonces $x$ y $x+v$ producen las mismas mediciones, por lo que no podemos distinguir entre ambas imágenes. En el ejercicio 4, $A_2$ y $A_4$ tienen núcleos de dimensiones 49 y 25, respectivamente, mientras que $A_7$ tiene rango 64.

**Mínimos cuadrados → solución frente al ruido.** Cuando hay ruido, generalmente $b$ no pertenece a $\operatorname{Col}(A)$, por lo que buscamos minimizar $\|Ax-b\|^2$. Geométricamente, $A\hat{x}$ es la proyección ortogonal de $b$ sobre $\operatorname{Col}(A)$ y el residuo cumple $A^T(b-A\hat{x})=0$ (ejercicio 7). La componente del ruido que pertenece a ese espacio se incorpora a la reconstrucción, mientras que la ortogonal queda en el residuo.

**Regularización → control del ruido.** Para obtener reconstrucciones más adecuadas se agrega el término $\lambda\|Dx\|^2$, que penaliza diferencias entre píxeles vecinos y supone que la imagen es relativamente suave. En el ejercicio 10, $\lambda=35$ redujo el granulado manteniendo los detalles, aunque una regularización excesiva puede difuminar los bordes.

**Métodos numéricos → cálculo de la solución.** Los métodos del ejercicio 8 permiten calcular las reconstrucciones: mínimos cuadrados mediante `lstsq`, Newton y gradiente descendente. Aunque siguen caminos distintos, pueden llegar a la misma solución del problema planteado.
