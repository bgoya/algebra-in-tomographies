 Se considera un *problema inverso* porque el problema no se define en cómo medir mejor al objeto, sino en cómo hacer las mediciones más acertadas al escaneo ideal.

En la realidad, partimos de una medición $\vec{b}$, que se obtiene tras escanear un vector $\hat{x}$ con la matriz de direcciones $A$, esto es, $A\hat{x} = \vec{b}$.

Vimos que $\hat{x} = \vec{x} + \vec{v}$, siendo $\vec{x}$ el escaneo ideal, mientras que $\vec{v}$ es un vector que representa las diferencias entre este vector ideal y la medicion realizada, las cuales son imperceptibles para el escaner. Estos cumplen $A\vec{x} = 0$ y $A\vec{v} = 0$.

Continuando el desarrollo, vimos que $A\hat{x} = A(\vec{x} + \vec{v}) = \vec{b} + 0 = \vec{b}$. Es por ello que este error es imperceptible para el escaner.

Se comprobo que las formas de reducir este error, esto es, aproximar mejor a la medicion verdadera era mediante minimos cuadrados o SVD.

### Integración de conceptos

**Problema inverso → modelo de reconstrucción.** El problema se plantea como $b=Ax$, donde $x$ representa la imagen, $A$ describe las proyecciones y $b$ contiene las mediciones. Es un problema inverso porque conocemos las mediciones y buscamos recuperar la imagen.

**Espacio nulo → unicidad de la solución.** La reconstrucción es única si $\operatorname{Nul}(A)=\{0\}$, equivalente a que $\operatorname{rango}(A)=n^2$. Si existe un vector $v\neq 0$ tal que $Av=0$, entonces $x$ y $x+v$ producen las mismas mediciones, por lo que no podemos distinguir entre ambas imágenes. En el ejercicio 4, $A_2$ y $A_4$ tienen núcleos de dimensiones 49 y 25, respectivamente, mientras que $A_7$ tiene rango 64.

**Mínimos cuadrados → solución frente al ruido.** Cuando hay ruido, generalmente $b$ no pertenece a $\operatorname{Col}(A)$, por lo que buscamos minimizar $\|Ax-b\|^2$. Geométricamente, $A\hat{x}$ es la proyección ortogonal de $b$ sobre $\operatorname{Col}(A)$ y el residuo cumple $A^T(b-A\hat{x})=0$ (ejercicio 7). La componente del ruido que pertenece a ese espacio se incorpora a la reconstrucción, mientras que la ortogonal queda en el residuo.

**Regularización → control del ruido.** Para obtener reconstrucciones más adecuadas se agrega el término $\lambda\|Dx\|^2$, que penaliza diferencias entre píxeles vecinos y supone que la imagen es relativamente suave. En el ejercicio 10, $\lambda=35$ redujo el granulado manteniendo los detalles, aunque una regularización excesiva puede difuminar los bordes.

**Métodos numéricos → cálculo de la solución.** Los métodos del ejercicio 8 permiten calcular las reconstrucciones: mínimos cuadrados mediante `lstsq`, Newton y gradiente descendente. Aunque siguen caminos distintos, pueden llegar a la misma solución del problema planteado.
